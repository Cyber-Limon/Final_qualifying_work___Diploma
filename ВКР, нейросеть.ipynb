# ВЫПУСКНАЯ КВАЛИФИКАЦИОННАЯ РАБОТА (ВКР)
## Математическое и программное обеспечение для прогнозирования свойств медицинских композиционных материалов

### Импорт библиотек:
- "pandas"     -
- "sklearn"    - :
- "tensorflow" - :

```python
import numpy  as np
import pandas as pd

from tensorflow              import keras
from sklearn.preprocessing   import StandardScaler
from tensorflow.keras.layers import Dense, Dropout
from sklearn.model_selection import train_test_split
```

### Загрузка данных:
- "compound"   - массив составов сплавов
- "properties" - массив свойств сплавов

```python
compound   = pd.read_excel("ВКР, датасет.xlsx", usecols="B:V")
compound   = compound.values.tolist()
compound   = np.array(compound)

properties = pd.read_excel("ВКР, датасет.xlsx", usecols="W:AD")
properties = properties.values.tolist()
properties = np.array(properties)
```

### Разделение данных на тренировочную и тестовую выборки:
- "X_train" - тренировочные составы
- "X_test"  - тестовые составы
- "y_train" - тренировочные свойства
- "y_test"  - тестовые свойства

```python
X_train, X_test, y_train, y_test = train_test_split(
    compound,
    properties,
    test_size=0.2,
    random_state=1107,
    shuffle=True
)
```

### Нормализация данных

```python
scaler_X = StandardScaler()
X_train  = scaler_X.fit_transform(X_train)
X_test   = scaler_X.transform(X_test)

scaler_Y = StandardScaler()
y_train  = scaler_Y.fit_transform(y_train)
y_test   = scaler_Y.transform(y_test)
```

### Архитектура модели

```python
model = keras.Sequential([
    keras.Input(shape=(X_train.shape[1],)),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(128, activation='relu'),
    Dropout(0.4),
    Dense(64,  activation='relu'),
    Dropout(0.5),
    Dense(32,  activation='relu'),
    Dense(y_train.shape[1])
])
```

### Обучение модели

```python
model.compile(optimizer='adam', loss='mse', metrics=['mae'])

model.summary()

log = model.fit(
    X_train, y_train,
    batch_size=32,
    epochs=300,
    validation_split=0.2,
    verbose=1
)
```

### Тестирование модели

```python
model.evaluate(X_test, y_test, verbose=1)
```
